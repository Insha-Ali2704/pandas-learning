```python
import pandas as pd
```

Series:1D labelled array that can hold any data type ie a single column

DataFrame:A table

```python
data=[100.1,102.2,104.3]
series=pd.Series(data)#constructor
print(series)
data=[1000,102,204,90,200]
series=pd.Series(data)#constructor
print(series)
print("\nTo print numbers greater than 200:\n",series[series>200])
data=["A","B","C"]#indexing apni marzi se dedo
series=pd.Series(data,index=["#1","#2","#3"])#constructor
print(series)
data=[True,False,True]#indexing apni marzi se dedo
series=pd.Series(data,index=["a","b","c"])#constructor
print(series.loc["b"])#to print any specific element
#OR
print(series.iloc[1])
#to change value of element 3
series.loc["c"]=False
print(series)
```

```python
calories = {"Day 1": 1750, "Day 2": 2110, "Day 3": 1700}
series = pd.Series(calories)#you dont need to paas index
print(series)
print("\nCalories consumed on Day 2:\n",series.loc["Day 2"])
#to update calory intake on Day 2
series.loc["Day 2"] += 1000
print("\nUpdated:\n",series)
#to print specifc / conditional data
print("\nDays where I consumed calories<2000:\n",series[series<2000])
```

```python
data={"Name":["Spongebob","Patrick","Squidward"],
      "Age":[30,35,50]
      }
#you can also write df=pd.DataFrame(data)
df=pd.DataFrame(data,index=["Employee 1","Employee 2","Employee 3"])
print(df)
#to access specific data
print(df.loc["Employee 1"])
#OR
print(df.iloc[2])
#Add a new column
df["Job"]=["Cook","N/A","Cashier"]
print(df)
#Add a new row
new_row=pd.DataFrame([{"Name":"Sandy","Age":28,"Job":"Engineer"}],index=["Employee 4"])
df=pd.concat([df,new_row])
print(df)
#Add a new rows
new_rows=pd.DataFrame([{"Name":"Eugene","Age":60,"Job":"Manager"},
                      {"Name":"Jack","Age":30,"Job":"Receptionist"}
                      ],index=["Employee 5","Employee 6"])
df=pd.concat([df,new_rows])
print(df)
```

FILTERING

```python
#filtering:keeping the rows that match a condition
df=pd.read_csv("dataset.csv")
adult=df[df["Age"]>=18]
#df[condition]
print(adult,"\n")
#to print data of adult females
adult_female=df[(df["Age"]>=18) & (df["Gender"]=="Female")]
print(adult_female)
```

Group By

```python
df=pd.read_csv("dataset.csv")
group=df.groupby("Gender")
print("Avg age of each gender type : \n",group["Age"].mean())
print("Sum of weekend extra hours of each gender type : \n",group["Weekend_Extra_Hours"].sum())
print("Min of weekend extra hours of each gender type : \n",group["Weekend_Extra_Hours"].min())
print("Max of weekend extra hours of each gender type : \n",group["Weekend_Extra_Hours"].max())
print("Count of weekend extra hours of each gender type : \n",group["Weekend_Extra_Hours"].count())
```

DATA CLEANING

```python
#the process of fixing/removing:incorrect,incomplete or irrelevant data.
df=pd.read_csv("dataset.csv")
#1.Drop irrelevant columns
print("1.Drop irrelevant columns")
df=df.drop(columns=["Sleep_Duration_Hours","Sleep_Quality_Score"])
print(df)
#2.Handle missing data
print("2.Handle missing data:to remove")
df=df.dropna(subset=["Academic_Level"])
print(df.to_string())
print("2.Handle missing data:to fill/replace values")
df=df.fillna({"Academic_Level":"None"})
print(df.to_string())
#3.Fix inconsistent values
print("Inconsistent values:\n")
df["Academic_Level"]=df["Academic_Level"].replace({"Postgraduate":"PG","Undergraduate":"UG"})
print(df.to_string())
#4.Standardize text
print("Standardize text\n")
df["Primary_Platform"]=df["Primary_Platform"].str.lower()
print(df.to_string())
#5.Fix data types
print("Fix data types\n")
df["Daily_Usage_Hours"]=df["Daily_Usage_Hours"].astype(int)
print(df.to_string())
#6.Remove duplicates
print("Remove duplicates:\n")
df=df.drop_duplicates()
print(df.to_string())
```
